## List your clusters, then shut them all down

* This notebook is adopted from Richard Signell's GitHub: [link](https://github.com/MoeinDst/MAGICA-School-2026/blob/main/list_and_shutdown_all_clusters.ipynb).


* **Note**!
    this notebook works only for cloud computing clusters -- such as AWS or Coiled and ***not*** for local cluster! For the local cluster one should simply run *clinet.close()* and *cluster.close()*.

```python
from dask_gateway import Gateway
from dask.distributed import Client
gateway = Gateway()
```

### List clusters

```python
gateway.list_clusters()
```

### Shut them all down

```python
for cluster in gateway.list_clusters():
    gateway.stop_cluster(cluster_name=cluster.name)
    print(f'stopped {cluster.name}')
```

```python
### double-checking to see if it took effect:
gateway.list_clusters()
```
